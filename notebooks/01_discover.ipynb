```python
import xml.etree.ElementTree as ET
from collections import Counter
from pathlib import Path

XML_DIR = Path("data/raw/xml")
print(len(list(XML_DIR.glob("*.xml"))), "fichiers XML")
```

## 2.3 Charger une fiche

```python
root = ET.parse(XML_DIR / "F33902.xml").getroot()
print(root.tag)
print(root.attrib)
```

```python
for child in root:
    print(child.tag)
```

## 2.4 Les métadonnées

```python
DC = "{http://purl.org/dc/elements/1.1/}"

print("ID      :", root.get("ID"))
print("type    :", root.get("type"))
print("titre   :", root.findtext(DC + "title"))
print("date    :", root.findtext(DC + "date"))
print("adresse :", root.get("spUrl"))
```

## 2.5 Le texte

```python
text = root.find("Texte")
ET.indent(text)
print(ET.tostring(text, encoding="unicode")[:1500])
```

```python
for paragraph in text.findall("Paragraphe"):
    print("".join(paragraph.itertext()))
    print()
```

## 2.6 Les liens

```python
for link in root.iter("LienIntra"):
    print(link.get("LienID"), "|", link.get("type"), "|", link.text)
```

```python
for definition in root.findall("Definition"):
    print(definition.get("ID"), ":", definition.findtext("Titre"))
```

```python
for question in root.findall("QuestionReponse"):
    print(question.get("ID"), ":", question.findtext("Titre"))
```

## 2.7 Une fiche longue : F2258

```python
long_root = ET.parse(XML_DIR / "F2258.xml").getroot()
print(long_root.findtext(DC + "title"))
print(long_root.get("type"))

tags = []
for child in long_root:
    if not child.tag.startswith("{"):
        tags.append(child.tag)
print(Counter(tags))
```

```python
for chapter in long_root.find("Texte").findall("Chapitre"):
    title = "".join(chapter.find("Titre").itertext()).strip()
    print(title)
    for case in chapter.iter("Cas"):
        case_title = "".join(case.find("Titre").itertext()).strip()
        print("    cas :", case_title)
```

```python
targets = []
for link in long_root.iter("LienInterne"):
    targets.append(link.get("LienPublication")[0])
print(Counter(targets))

for link in long_root.iter("LienInterne"):
    if link.get("LienPublication").startswith("F"):
        print(link.get("LienPublication"), "|", link.text)
        break
```

```python
for reference in long_root.findall("Reference")[:5]:
    print(reference.findtext("Titre"))
```

```python
situations_root = ET.parse(XML_DIR / "F2382.xml").getroot()
print(situations_root.findtext(DC + "title"))
print("Texte :", situations_root.find("Texte"))
for situation in situations_root.find("ListeSituations").findall("Situation"):
    print("situation :", situation.findtext("Titre").strip())
```

## 2.8 Le dossier N510

```python
folder = root.find("DossierPere")
print(folder.get("ID"), ":", folder.findtext("Titre"))
for subfolder in folder.findall("SousDossier"):
    print("  ", subfolder.findtext("Titre"), ":", len(subfolder.findall("Fiche")), "fiches")
```

```python
types = []
for path in XML_DIR.glob("F*.xml"):
    fiche = ET.parse(path).getroot()
    if fiche.find("DossierPere[@ID='N510']") is not None:
        types.append(fiche.get("type"))

print(len(types), "fiches dans N510")
print(Counter(types))
```
