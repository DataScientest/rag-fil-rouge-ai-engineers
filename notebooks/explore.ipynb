```python
def show(question: str, k: int = 5) -> None:
    """Affiche doc_id, section et début du texte des k résultats."""
    print("#", question)
    for doc in search(question, k):
        print(f"  {doc.metadata['doc_id']:7} {doc.metadata['section_path'][:70]:70} | {doc.page_content[:60]!r}")


for q in ["Combien de jours de congé pour le décès de mon père ?",
          "Mon employeur peut-il refuser mes dates de congés payés ?",
          "Que dit l'article L3142-4 du Code du travail ?"]:
    show(q)
```
